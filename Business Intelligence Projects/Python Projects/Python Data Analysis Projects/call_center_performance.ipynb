 VISUALIZATION DASHBOARD

```python
import plotly.express as px
```

Calls by Department

```python
dept_calls = df.groupby('Department')['Call_ID'].count().reset_index()

fig = px.bar(
    dept_calls,
    x='Department',
    y='Call_ID',
    title='Calls by Department'
)

fig.show()
```

Calls by Shift

```python
fig = px.pie(
    df,
    names='Shift',
    title='Calls by Shift'
)

fig.show()
```

Daily Call Trend

```python
daily_calls = df.groupby('Call_Date')['Call_ID'].count().reset_index()

fig = px.line(
    daily_calls,
    x='Call_Date',
    y='Call_ID',
    title='Daily Call Trend'
)

fig.show()
```

Customer Satisfaction
Satisfaction by Department

```python
sat = df.groupby('Department')['Customer_Satisfaction'].mean().reset_index()

fig = px.bar(
    sat,
    x='Department',
    y='Customer_Satisfaction',
    title='Customer Satisfaction by Department'
)

fig.show()
```

Satisfaction vs Wait Time

```python
fig = px.scatter(
    df,
    x='Wait_Time_Seconds',
    y='Customer_Satisfaction',
    color='Department',
    title='Wait Time vs Satisfaction'
)

fig.show()
```

Agent Performance Dashboard
Top Agents by Rating

```python
agent_perf = df.groupby('Agent_ID')['Agent_Rating'].mean().reset_index()

top_agents = agent_perf.sort_values(
    by='Agent_Rating',
    ascending=False
).head(10)

fig = px.bar(
    top_agents,
    x='Agent_ID',
    y='Agent_Rating',
    title='Top 10 Agents'
)

fig.show()
```

Calls Handled Per Agent

```python
fig = px.box(
    df,
    x='Shift',
    y='Calls_Handled_Per_Day',
    title='Calls Handled Per Day'
)

fig.show()
```

Resolution Analytics
Resolution Status Distribution

```python
fig = px.pie(
    df,
    names='Resolution_Status',
    title='Resolution Status'
)

fig.show()
```

First Call Resolution Rate

```python
fcr = df.groupby('Department')['First_Call_Resolution'].mean().reset_index()

fig = px.bar(
    fcr,
    x='Department',
    y='First_Call_Resolution',
    title='First Call Resolution Rate'
)

fig.show()
```

Escalation Analytics
Escalation by Issue Category

```python
esc = df.groupby('Issue_Category')['Escalation'].mean().reset_index()

fig = px.bar(
    esc,
    x='Issue_Category',
    y='Escalation',
    title='Escalation Rate by Issue'
)

fig.show()
```

Wait Time Analytics
Average Wait Time by Department

```python
wait = df.groupby('Department')['Wait_Time_Seconds'].mean().reset_index()

fig = px.bar(
    wait,
    x='Department',
    y='Wait_Time_Seconds',
    title='Average Wait Time'
)

fig.show()
```

Call Type Analysis

```python
fig = px.histogram(
    df,
    x='Call_Type',
    title='Call Type Distribution'
)

fig.show()
```
