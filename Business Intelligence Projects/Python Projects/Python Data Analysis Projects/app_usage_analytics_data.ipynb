 VISUALIZATION DASHBOARD

```python
import plotly.express as px
```

```python
total_users = df['User_ID'].nunique()
total_sessions = df['Session_ID'].nunique()
avg_duration = df['Session_Duration_Min'].mean()
total_revenue = df['Purchase_Amount'].sum()
crash_rate = df['Crash_Flag'].mean() * 100
retention_rate = df['Retention_Flag'].mean() * 100
```

User Engagement Dashboard

Session Duration Distribution

```python
px.histogram(df, x='Session_Duration_Min')
```

Active vs Idle Time

```python
px.scatter(
    df,
    x='Active_Time_Min',
    y='Idle_Time_Min',
    color='Device_Type'
)
```

Screens Viewed by Device

```python
px.box(df, x='Device_Type', y='Screens_Viewed')
```

Revenue Dashboard

Revenue by Country

```python
country_rev = df.groupby('Country')['Purchase_Amount'].sum().reset_index()

px.bar(country_rev,
       x='Country',
       y='Purchase_Amount')
```

In-App Purchase Conversion

```python
purchase_rate = df['In_App_Purchase'].mean() * 100
```

Revenue by Device Type

```python
px.pie(df,
       names='Device_Type',
       values='Purchase_Amount')
```

Crash Analytics Dashboard

Crash Rate by App Version

```python
crash_version = df.groupby('App_Version')['Crash_Flag'].mean().reset_index()

px.line(crash_version,
        x='App_Version',
        y='Crash_Flag')
```

Crash by OS

```python
px.bar(df.groupby('OS')['Crash_Flag'].mean().reset_index(),
       x='OS',
       y='Crash_Flag')
```

Retention Analytics

Retention by Country

```python
ret = df.groupby('Country')['Retention_Flag'].mean().reset_index()

px.bar(ret,
       x='Country',
       y='Retention_Flag')
```

Retention vs Session Duration

```python
px.scatter(
    df,
    x='Session_Duration_Min',
    y='Retention_Flag'
)
```

Advertisement Analytics

Ad Clicks by Country

```python
ads = df.groupby('Country')['Ad_Clicks'].sum().reset_index()

px.bar(ads,
       x='Country',
       y='Ad_Clicks')
```
