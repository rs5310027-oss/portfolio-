```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv("C:/Users/hp/OneDrive/Desktop/Project github/New folder/New folder/subscription_analysis_data.csv")

# Convert dates
df['Start_Date'] = pd.to_datetime(df['Start_Date'])
df['End_Date'] = pd.to_datetime(df['End_Date'])
df['Last_Login_Date'] = pd.to_datetime(df['Last_Login_Date'])
```

```python
pivot_churn = pd.pivot_table(df,
                             values='Churn_Flag',
                             index='Subscription_Type',
                             aggfunc='mean') * 100
```

```python
pivot_revenue = pd.pivot_table(df,
                               values='Monthly_Fee',
                               index='Subscription_Type',
                               aggfunc='sum')
```

```python
pivot_city = pd.pivot_table(df,
                            values='Satisfaction_Score',
                            index='City',
                            aggfunc='mean')
```

```python
pivot_usage = pd.pivot_table(df,
                             values='Usage_Hours_Per_Month',
                             index='Churn_Flag',
                             aggfunc='mean')
```

```python
total_customers = df['Customer_ID'].nunique()
churn_rate = df['Churn_Flag'].mean() * 100
avg_revenue = df['Monthly_Fee'].mean()
avg_usage = df['Usage_Hours_Per_Month'].mean()
avg_satisfaction = df['Satisfaction_Score'].mean()
```

```python
plt.figure()
df['Churn_Flag'].value_counts().plot(kind='bar')
plt.title("Churn Distribution")
plt.show()
```

```python
plt.figure()
sns.barplot(x='Subscription_Type', y='Churn_Flag', data=df)
plt.title("Churn Rate by Plan")
plt.show()
```

```python
plt.figure()
sns.boxplot(x='Churn_Flag', y='Tenure_Months', data=df)
plt.title("Tenure vs Churn")
plt.show()
```

```python
feature_cols = [col for col in df.columns if "Feature_" in col]

feature_impact = df.groupby('Churn_Flag')[feature_cols].mean().T
feature_impact.plot(kind='bar', figsize=(12,6))
plt.title("Feature Usage vs Churn")
plt.show()
```

```python
cat_cols = ['Gender', 'City', 'Subscription_Type', 'Payment_Method', 'Auto_Renew']

le = LabelEncoder()

for col in cat_cols:
    df[col] = le.fit_transform(df[col].astype(str))
```

```python
df.fillna(0, inplace=True)
```
