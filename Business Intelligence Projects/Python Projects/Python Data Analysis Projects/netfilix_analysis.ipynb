```python
pivot1 = pd.pivot_table(df,
                        values='Content_ID',
                        index='Genre',
                        columns='Type',
                        aggfunc='count')
print(pivot1)
```

```python
plt.figure()
df['Genre'].value_counts().plot(kind='bar')
plt.title("Content Distribution by Genre")
plt.show()
```

```python
plt.figure()
plt.scatter(df['IMDb_Score'], df['RottenTomatoes_%'])
plt.xlabel("IMDb Score")
plt.ylabel("Rotten Tomatoes %")
plt.title("Ratings Comparison")
plt.show()
```

```python
plt.figure()
plt.scatter(df['Budget_Million'], df['Revenue_Million'])
plt.xlabel("Budget")
plt.ylabel("Revenue")
plt.title("Budget vs Revenue")
plt.show()
```

```python
top_genre = df.groupby('Genre')['Popularity_Index'].mean().sort_values(ascending=False)

plt.figure()
top_genre.plot(kind='bar')
plt.title("Top Genres by Popularity")
plt.show()
```
