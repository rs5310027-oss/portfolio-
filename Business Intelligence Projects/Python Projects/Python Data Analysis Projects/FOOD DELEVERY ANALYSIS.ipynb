```python
# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
#Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
# Convert datetime
df['Order_Date_and_Time'] = pd.to_datetime(df['Order_Date_and_Time'], format='%d-%m-%Y %H:%M')
df['Delivery Date and Time'] = pd.to_datetime(df['Delivery Date and Time'], format='%d-%m-%Y %H:%M')
```

```python
# Delivery time
df['Delivery_Time_Min'] = (
    df['Delivery Date and Time'] - df['Order_Date_and_Time']
).dt.total_seconds() / 60


print("\nData Cleaned Successfully")
```

```python
# KPI
print("Total Revenue:", df['Order_Value'].sum())
print("Average Order Value:", df['Order_Value'].mean())
print("Average Delivery Time:", df['Delivery_Time_Min'].mean())
```

```python
# -------------------------------
# 3. KPI Reporting
# -------------------------------

total_orders = df.shape[0]
total_revenue = df['Order_Value'].sum()
avg_order_value = df['Order_Value'].mean()

print("\n===== KPI REPORT =====")
print(f"Total Orders: {total_orders}")
print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Average Order Value: ₹{avg_order_value:.2f}")
```


Exploratory Data Analysis

```python
 # Revenue by Payment Method
payment_analysis = df.groupby('Payment_Method')['Order_Value'].sum()
print("\nRevenue by Payment Method")
print(payment_analysis)

# Top Restaurants
top_restaurants = df.groupby('Restaurant')['Order_Value'].sum().sort_values(ascending=False).head(10)
print("\nTop 10 Restaurants")
print(top_restaurants)
```

Data Visualization

```python
# Revenue Distribution
plt.figure(figsize=(10,5))
sns.histplot(df['Order_Value'], bins=30)
plt.title("Order Value Distribution")
plt.xlabel("Order Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# Payment Method Analysis
plt.figure(figsize=(8,5))
sns.countplot(x='Payment_Method', data=df)
plt.title("Payment Method Usage")
plt.show()
```

```python
import plotly.express as px
```

```python
# Revenue by Payment Method
fig1 = px.pie(
    df,
    names='Payment_Method',
    values='Order_Value',
    title='Revenue by Payment Method'
)
fig1.show()
```

```python
# Revenue Trend
daily_revenue = df.groupby(df['Order_Date_and_Time'].dt.date)['Order_Value'].sum().reset_index()

fig3 = px.line(
    daily_revenue,
    x='Order_Date_and_Time',
    y='Order_Value',
    title='Daily Revenue Trend'
)
fig3.show()

print("\nDashboard Analysis Complete")
```
